## 2260. Minimum Consecutive Cards to Pick Up

You are given an integer array `cards` where `cards[i]` represents the value of the `ith` card. A pair of cards are matching if the cards have the same value.

Return the <b><i>minimum</b> number of <b>consecutive</b> cards you have to pick up to have a pair of <b>matching cards</b> among the picked cards.</i> If it is impossible to have matching cards, return `-1`.

<b>Example 1:</b>

Input: cards = [3,4,2,3,4,7]

Output: 4

Explanation: We can pick up the cards [3,4,2,3] which contain a matching pair of cards with value 3. Note that picking up the cards [4,2,3,4] is also optimal.

<b>Example 2:</b>

Input: cards = [1,0,5,3]

Output: -1

Explanation: There is no way to pick up a set of consecutive cards that contain a pair of matching cards.

<b>Constraints:</b>

- `1 <= cards.length <= 10^5`
- `0 <= cards[i] <= 10^6`

```python
from typing import List

def minimumCardPickup(cards: List[int]) -> int:
    last_seen = {}
    min_len = float('inf')

    for index, card in enumerate(cards):
        if card in last_seen:
            min_len = min(min_len, index - last_seen[card] + 1)
        last_seen[card] = index

    return min_len if min_len != float('inf') else -1
```

```python
minimumCardPickup([3,4,2,3,4,7])
```

```python
minimumCardPickup([1,0,5,3])
```

```python
minimumCardPickup([5,9,4,1,9])
```
