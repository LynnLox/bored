## 2900. Longest Unequal Adjacent Groups Subsequence I

You are given a string array `words` and a binary array `groups` both of length n, where `words[i]` is associated with `groups[i]`.

Your task is to select the <b>longest alternating subsequence</b> from words. A subsequence of `words` is alternating if for any two consecutive strings in the sequence, their corresponding elements in the binary array groups differ. Essentially, you are to choose strings such that adjacent elements have non-matching corresponding bits in the groups array.

Formally, you need to find the longest subsequence of an array of indices `[0, 1, ..., n - 1]` denoted as `[i0, i1, ..., ik-1]`, such that `groups[ij] != groups[ij+1]` for each `0 <= j < k - 1 `and then find the words corresponding to these indices.

Return <i>the selected subsequence. If there are multiple answers, return <b>any</b> of them.</i>

<b>Note:</b> The elements in `words` are distinct.

<b>Example 1:</b>
- `words -> | e | a | b |`
- `groups-> | 0 | 0 | 1 |`

1. Order should be alternating 1,0,1,0,1,..... or 0,1,0,1,0,......
2. The original order of words must be preserved. So, skip elements if needed, but every adjacent pair in the selected subsequence must have different group values.

Which means you can't do this ❌

`words -> | e | b | a |`

`groups-> | 0 | 1 | 0 |`

Correct Way ✅:
words -> | e | b |
groups-> | 0 | 1 |

OR 

`words -> | a | b |`

`groups-> | 0 | 1 |`

<b>Example 2:</b>

`words -> | a | b | c | d |`

`groups-> | 1 | 0 | 1 | 1 |`

Start with "a" (group 1)

Next is "b" (group 0) → ✅ different from 1

Next is "c" (group 1) → ✅ different from 0

Next is "d" (group 1) → ❌ same as previous group 1, so we stop here

Longest alternating sequence 101 i.e. abc

1 (i.e. d) at the end, not fully alternating.

```python
from typing import List

def getLongestSubsequence(words: List[str], groups: List[int]) -> List[str]:
        result = [words[0]]
        last_group = groups[0]
        for i,word in enumerate(words):
            if groups[i]!= last_group:
                result.append(word)
                last_group = groups[i]

        return result if len(result) >= 1 else []
```

```python
print(getLongestSubsequence(["a", "b", "c", "d"], [1, 0, 1, 1]))
```

```python
print(getLongestSubsequence(["e","a","b"], [0, 0, 1]))
```

```python
print(getLongestSubsequence(["e","a","b"], [0, 0, 0]))
```
