## 2901. Longest Unequal Adjacent Groups Subsequence II

```python
def hamming_distance(s1:str, s2:str):   # sad and mad
    count = 0
    for ch1, ch2 in zip(s1, s2): # zip(s1,s2) => [('s','m'),('a','a'),('d','d')]
        if ch1 != ch2:
            count += 1
    return count
    # return sum(ch1!=ch2 for ch1,ch2 in zip(s1,s2)) # sum(true, f, f)



```

```python
from typing import List

def getLongestSubsequence(words: List[str], groups: List[int]) -> List[str]:
    n = len(words)
    dp = [[word] for word in words] # dp[i] stores longest subsequence ending at i

    for i in range(n):
        for j in range(i):
            if groups[i]!=groups[j] and len(words[i])==len(words[j]) and hamming_distance(words[i],words[j]) == 1:
                if len(dp[j])+1 > len(dp[i]):
                    dp[i] = dp[j] + [words[i]]
    
    return max(dp, key=len)

```

```python
getLongestSubsequence(['bad','sad','mad'],[1,2,2])
```
