## 3024. Type of Triangle

You are given a <b>0-indexed</b> integer array `nums` of size `3` which can form the sides of a triangle.

* A triangle is called <b>equilateral</b> if it has all sides of equal length.
* A triangle is called <b>isosceles</b> if it has exactly two sides of equal length.
* A triangle is called <b>scalene</b> if all its sides are of different lengths.

Return a string representing the type of triangle that can be formed or "none" if it cannot form a triangle.

```python
from typing import List


def triangleType(nums: List[int]):
    nums.sort()

    if nums[0] + nums[1] <= nums[2]:
        return "none"
    elif nums[0] == nums[2]:
        return "equilateral"
    elif nums[0]== nums[1] or nums[1] == nums[2]:
        return "isosceles"
    else:
        return "scalene"
```

```python
triangleType([8,4,2])
```

```python
triangleType([3,4,5])
```

```python
triangleType([3,3,3])
```

```python
triangleType([8,5,5])
```
